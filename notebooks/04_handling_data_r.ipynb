```r
library(tidyverse)
```

```r
subway <- read_csv(
  "../dataset/raw/seoul-subway-2025.csv",
  locale = locale(encoding = "EUC-KR")
)
```

```r
subway <- subway %>%
  filter(!is.na(수송일자)) %>%
  select(-연번) %>%
  rename(역 = 역명, 구분 = 승하차구분)

nrow(subway)
count(subway, 구분)   # 승차와 하차가 같은 수인가
```

```r
glimpse(subway)
```

```r
range(subway$수송일자)
```

```r
View(subway)
```

```r
subway_long <- subway %>%
  pivot_longer(
    cols = `06시이전`:`24시이후`,
    names_to = "시간대",
    values_to = "인원"
  )
```

```r
# 수정: 열 이름을 작은따옴표('...')로 쓰면 문자열이 되어 NA가 나온다 → 백틱(`...`)으로 바꾸고, 없는 열 '0시이전'을 '06시이전'으로 고침
subway %>%
    summarise(t1.mean = mean(`06시이전`),
    t2.mean = mean(`06-07시간대`),
    t3.mean = mean(`07-08시간대`))
```

```r
subway <- subway %>%
  mutate(날짜 = as_date(수송일자))

sum(is.na(subway$날짜))   # 0이어야 한다
range(subway$날짜)
```

```r
subway <- subway %>%
  mutate(하루 = rowSums(
    across(`06시이전`:`24시이후`)))

daily <- subway %>%
  filter(구분 == "승차") %>%
  group_by(날짜) %>%
  summarise(승차 = sum(하루))

nrow(daily)   # 365
```

```r
View(head(daily))
```

```r
daily <- daily %>%
  mutate(
    요일 = wday(날짜, label = TRUE),
    주말 = wday(날짜) %in% c(1, 7)
  )
# wday(): 1 = 일요일, 7 = 토요일
```

## 기상 데이터 정리하기

```r
weather <- read_csv(
  "../dataset/raw/seoul-temp-raining-2025.csv",
  locale = locale(encoding = "EUC-KR")
) %>%
  select(날짜 = 일시,
         기온 = `평균기온(°C)`,
         강수량 = `일강수량(mm)`) %>%
  mutate(날짜 = as_date(날짜))

nrow(weather)   # 365
```

```r
weather <- weather %>%
  mutate(강수량 = replace_na(강수량, 0),
         비 = 강수량 > 0)
```

```r
head(weather)
```

```r
# 수정: subway(역 단위)가 아니라 daily(하루 단위, 365행)에 날씨를 붙인다. 기존 코드는 9·11번 셀에서 만든 daily를 역 단위 표로 덮어써서 승차·주말 열이 사라졌다
nrow(daily)                       # 조인 전: 365

daily <- daily %>%
  left_join(weather, by = "날짜")

nrow(daily)                       # 조인 후: 365
```

```r
# 수정: daily가 하루 단위 표가 되어 시간대 열·구분 열이 없으므로, 열 정리 대신 구조만 확인한다
glimpse(daily)
```
