```r
subway <- read_csv(
  "../dataset/raw/seoul-subway-2025.csv",
  locale = locale(encoding = "EUC-KR")
)
```

```r
sum(is.na(subway$수송일자))   # 빈 행의 수
filter(subway, is.na(수송일자))   # 수송일자가 결측인(NA) 행 확인
filter(subway, !is.na(수송일자))  # 수송일자가 결측이 아닌 행 확인
subway %>% filter(!is.na(수송일자))   # 수송일자가 결측이 아닌 행 확인 (직전 코드와 동일)

subway %>% filter(!is.na(수송일자)) %>% summarise(n=n())
subway |> filter(!is.na(수송일자)) %>% summarise(n=n())
```
